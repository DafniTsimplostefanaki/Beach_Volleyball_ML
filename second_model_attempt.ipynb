```python
#I encoded the columns that contain text in x_train so that the model can read the int values instead of text

X_encoded = pd.get_dummies(
    data,
    columns=['passing_type', 'attack_zone', 'defense_position', 'attack_type']
)

feature_names = [
    column for column in X_encoded.columns
    if column != 'match_id' and column != 'attack_outcome'
]

x_train = np.array(X_encoded[feature_names])
print(x_train.shape)
print(x_train)
```
